```python
with open("/Users/simonchervenak/Downloads/lings-anki.csv", "w") as f:
    writer = csv.DictWriter(f, fieldnames=['term', 'meaning1'])
    writer.writeheader()
    with open(file) as f:
        reader = csv.DictReader(f)
        for row in reader:
            nr = {'term': row['term'], 'meaning1': f'{pinyin.get(row['term'])} - {row['meaning1']}'}
            writer.writerow(nr)
```
